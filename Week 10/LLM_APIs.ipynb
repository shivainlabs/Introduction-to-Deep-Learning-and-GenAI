```python
from groq import Groq
client = Groq(api_key = GROQ_API_KEY)
MODEL = "openai/gpt-oss-120b"

```

```python
completion = client.chat.completions.create(
    model = MODEL,
    messages = [

      {
          'role': 'system',
          'content':'you are helpful teacher'
      },
      {
          'role':'user',
          'content':'In one sentence, explain API Keys'
      }
    ]

)
```

```python
def quick_ask(system_msg,user_msg):
  resp = client.chat.completions.create(
      model = MODEL,
      messages = [
          {'role':"system","content":system_msg},
          {"role":"user","content":user_msg}
      ]
  )

  return resp.choices[0].message.content
```

```python
question = "what is gravity"
print(quick_ask("you explain things to a 5 years old kid using simple words",question))
```

```python
def ask_groq(
    prompt,
    system = 'you are a helpful assistant',
    model = MODEL,
    temperature = 0.7, # higher means more dynamic response
    max_completion_tokens = 1200
):

  completion = client.chat.completions.create(
        model = MODEL,
        messages = [
            {'role':"system","content":system},
            {"role":"user","content":prompt}
        ],
        temperature = temperature,
        max_completion_tokens = max_completion_tokens
    )
  return completion.choices[0].message.content

print(ask_groq("Give me the fun one line fact about number zero"))

```

```python
stream = client.chat.completions.create(
    model = MODEL,
    messages = [{"role":"user",'content':"List 10 study tips for exams, in detail."}],
    stream = True,
)

for chunk in stream:
  piece = chunk.choices[0].delta.content
  if piece is not None:
    print(piece, end="", flush=True)

# Improve interactions, giving output chunk by chunk instead getting whole result first then gives
```
